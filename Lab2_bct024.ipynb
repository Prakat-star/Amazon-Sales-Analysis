One-Sample t_test


```python
import numpy as np
import scipy.stats as stats
scores=[78,72,74,76,73,79,77,71,70,74]
pmean=75
n=len(scores)
samplemean=np.mean(scores)
samplestd=np.std(scores,ddof=1)
tstat=(samplemean-pmean)/(samplestd/np.sqrt(n))
df=n-1
pvalue=2*stats.t.sf(abs(tstat),df)
print(f"sample Mean:{samplemean:.2f}")
print(f"t-statistic:{tstat:.2f}")
print(f"p-value:{pvalue:.4f}")
if pvalue<0.05:
  print("Reject the null hypothesis:The sample mean is significantly different")
else:
  print("Fail to reject null hypothesis:NO significant difference")
```

Two Sample t-test

```python
import numpy as np
import scipy.stats as stats
classa=[85,88,90,92,86,84,87,89,91,93]
classb=[78,80,82,79,81,77,83,76,75,80]
tstat,pvalue=stats.ttest_ind(classa,classb,equal_var=False)
print(f"t-statistic:{tstat:.2f}")
print(f"p-value:{pvalue:.4f}")
if pvalue<0.05:
  print("Reject the null hypothesis:The sample mean is significantly different")
else:
  print("Fail to reject null hypothesis:NO significant difference")
```

One sample z-test

```python
import numpy as np
import scipy.stats as stats
samplemean=495
pmean=500
pstd=10
n=40
zstat=(samplemean-pmean)/(pstd/np.sqrt(n))
pvalue=2*stats.norm.sf(abs(zstat))
print(f"z-statistic:{zstat:.2f}")
print(f"p-value:{pvalue:.4f}")
if pvalue<0.05:
  print("Reject the null hypothesis:The sample mean is significantly different")
else:
  print("Fail to reject null hypothesis:NO significant difference")
```

scenario: qn of z test

```python
import numpy as np
import scipy.stats as stats
mean1,std1,n1=495,10,50
mean2,std2,n2=500,12,60
zstat=(mean1-mean2)/((std1**2/n1+std2**2/n2)**0.5)
pvalue=2*stats.norm.sf(abs(zstat))
print(f"z-statistic:{zstat:.2f}")
print(f"p-value:{pvalue:.4f}")
if pvalue<0.05:
  print("Reject the null hypothesis:The sample mean is significantly different")
else:
  print("Fail to reject the null hypothesis:No significant difference")
```
