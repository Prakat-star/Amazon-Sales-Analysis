Merging Joining Concatenating

```python
import pandas as pd

df1=pd.DataFrame({'A':['A0','A1','A2','A3'],'B':['B0','B1','B2','B3'],'C':['C0','C1','C2','C3'],'D':['D0','D1','D2','D3']})
print(df1)
df2=pd.DataFrame({'A':['A4','A5','A6','A7'],'B':['B4','B5','B6','B7'],'C':['C4','C5','C6','C7'],'D':['D4','D5','D6','D7']},index=[4,5,6,7])

print(df2)
df3=pd.DataFrame({'A':['A8','A9','A10','A11'],'B':['B8','B9','B10','B11'],'C':['C8','C9','C10','C11'],'D':['D8','D9','D10','D11']},index=[8,9,10,11])

print(df3)
print(pd.concat([df1,df2,df3],axis=1))
left=pd.DataFrame({'A':['A0','A1','A2','A3'],'B':['B0','B1','B2','B3'],'key1':['K0','K1','K2','K3'],'key2':['K0','K1','K2','K3']})
right=pd.DataFrame({'key1':['K0','K1','K2','K3'],'key2':['K0','K1','K2','K3'],'C':['C0','C1','C2','C3'],'D':['D0','D1','D2','D3']})
print(pd.merge(left,right,how='outer',on=['key1','key2']))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
dataset=np.random.randn(25)
sns.rugplot(dataset)
xmin=dataset.min()-2
xmax=dataset.max()+2
xaxis=np.linspace(xmin,xmax,100)
url='http://en.wikipedia.org/wiki/Kernel density estimation#Pracitcal estimation of the bandwidth'
bandwidth=((4*dataset.std()**5)/(3*len(dataset)))**.2
kernell=[]
for datapoint in dataset:
  kernel=stats.norm(datapoint,bandwidth).pdf(xaxis)
  kernell.append(kernel)
  kernel=kernel/kernel.max()
  kernel=kernel*.4
  plt.plot(xaxis,kernel,color='grey',alpha=0.5)
plt.ylim(0,1)
```
